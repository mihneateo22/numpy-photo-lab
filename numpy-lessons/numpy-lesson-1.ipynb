## However, when dealing with floats, the exact number of elements in the array is not always predictable. For example, consider this:

```python
print(np.arange(0, 5/3, 1/3)) # depending on floating point errors, the max value is 4/3 or 5/3.
print(np.arange(0, 5/3, 0.333333333))
print(np.arange(0, 5/3, 0.333333334))
```

## For this reason, for floating points we are using linspace

```python
print(np.linspace(0, 5/3, 6)) # it gives 6 evenly spaced numbers between 0 and 5/3 INCLUDED, unlike arange or range in python
```

## You can also initialize a ndarray using a function (below)

```python
def my_function(z, y, x):
    return x + 10 * y + 100 * z

np.fromfunction(my_function, (3, 2, 10)) # THE FUNCTION IS GETTING CALLED ONLY ONCE
```

```python
x = np.arange(0, 1024, dtype = np.uint8) # 0-255 4 times stored in x
print(x, x.itemsize) # itemsize is the number of bytes of each item in the array. for example for a np.uint16 the itemsize = 2
print()
y = np.arange(0, 1000, dtype = np.uint8)
print(y, y.itemsize)
```
